```python
import math

# --- 1. FEATURE & TRAFFIC ASSUMPTIONS ---
feature_name = "Customer Support AI Assistant"  # @param {type:"string"}
daily_active_users = 5000  # @param {type:"integer"}
requests_per_user_per_day = 4  # @param {type:"number"}
avg_input_tokens = 1200  # @param {type:"integer"}
avg_output_tokens = 350  # @param {type:"integer"}
days_per_month = 30  # @param {type:"integer"}
# Peak traffic multiplier (e.g., 3.0 means peak minute traffic is 3x the daily average)
peak_traffic_multiplier = 3.0  # @param {type:"number"}

# --- 2. MODEL PRICING & CURRENCY (NOV 2026 RATES) ---
model_preset = "Gemini 3.8 Flash ($0.75 / $3.75 per 1M)"  # @param ["Gemini 3.8 Flash ($0.75 / $3.75 per 1M)", "Gemini 3.5 Flash-Lite ($0.30 / $2.50 per 1M)", "Gemini 4 Argon ($2.00 / $10.00 per 1M)", "Custom Pricing"]
custom_input_usd_per_1m = 1.00  # @param {type:"number"}
custom_output_usd_per_1m = 5.00  # @param {type:"number"}
usd_to_inr = 93.27  # @param {type:"number"}

# --- 3. PROVIDER RATE LIMITS & BUDGET ---
provider_rpm_limit = 2000  # @param {type:"integer"}
provider_tpm_limit = 1000000  # @param {type:"integer"}
monthly_budget_inr = 50000.0  # @param {type:"number"}
```

```python
def run_budget_estimator():
    # Resolve pricing per 1M tokens (USD)
    presets = {
        "Gemini 3.8 Flash ($0.75 / $3.75 per 1M)": (0.75, 3.75),
        "Gemini 3.5 Flash-Lite ($0.30 / $2.50 per 1M)": (0.30, 2.50),
        "Gemini 4 Argon ($2.00 / $10.00 per 1M)": (2.00, 10.00),
        "Custom Pricing": (custom_input_usd_per_1m, custom_output_usd_per_1m),
    }
    in_price_usd, out_price_usd = presets[model_preset]

    # Volume calculations
    daily_requests = daily_active_users * requests_per_user_per_day
    monthly_requests = daily_requests * days_per_month

    monthly_input_tokens = monthly_requests * avg_input_tokens
    monthly_output_tokens = monthly_requests * avg_output_tokens
    monthly_total_tokens = monthly_input_tokens + monthly_output_tokens

    # Cost calculations (USD -> INR)
    input_cost_usd = (monthly_input_tokens / 1_000_000) * in_price_usd
    output_cost_usd = (monthly_output_tokens / 1_000_000) * out_price_usd
    total_cost_usd = input_cost_usd + output_cost_usd

    input_cost_inr = input_cost_usd * usd_to_inr
    output_cost_inr = output_cost_usd * usd_to_inr
    total_cost_inr = total_cost_usd * usd_to_inr
    cost_per_request_inr = (
        total_cost_inr / monthly_requests if monthly_requests else 0
    )
    cost_per_dau_inr = (
        total_cost_inr / daily_active_users if daily_active_users else 0
    )

    # Rate limit calculations (Average vs. Peak Minute)
    minutes_per_day = 24 * 60
    avg_rpm = daily_requests / minutes_per_day
    peak_rpm = avg_rpm * peak_traffic_multiplier

    tokens_per_req = avg_input_tokens + avg_output_tokens
    avg_tpm = avg_rpm * tokens_per_req
    peak_tpm = peak_rpm * tokens_per_req

    # Status checks
    within_budget = total_cost_inr <= monthly_budget_inr
    within_rpm = peak_rpm <= provider_rpm_limit
    within_tpm = peak_tpm <= provider_tpm_limit

    budget_diff = monthly_budget_inr - total_cost_inr
    budget_pct = (
        (total_cost_inr / monthly_budget_inr) * 100 if monthly_budget_inr else 0
    )

    # --- REPORT OUTPUT ---
    print("=" * 62)
    print(f" LLM FEATURE BUDGET & CAPACITY REPORT: {feature_name}")
    print("=" * 62)

    print("\n1. TRAFFIC & TOKEN CONSUMPTION (MONTHLY)")
    print(f"   • Total Requests:       {monthly_requests:,.0f} req/month")
    print(f"   • Input Tokens:         {monthly_input_tokens:,.0f} tokens")
    print(f"   • Output Tokens:        {monthly_output_tokens:,.0f} tokens")
    print(f"   • Total Tokens:         {monthly_total_tokens:,.0f} tokens")

    print(f"\n2. ESTIMATED API COST (1 USD = ₹{usd_to_inr:.2f})")
    print(f"   • Model Tier:           {model_preset}")
    print(
        f"   • Input Cost:           ₹{input_cost_inr:,.2f} (${input_cost_usd:,.2f})"
    )
    print(
        f"   • Output Cost:          ₹{output_cost_inr:,.2f} (${output_cost_usd:,.2f})"
    )
    print(
        f"   • Total Monthly Cost:   ₹{total_cost_inr:,.2f} (${total_cost_usd:,.2f})"
    )
    print(f"   • Cost per Request:     ₹{cost_per_request_inr:.4f}")
    print(f"   • Cost per Active User: ₹{cost_per_dau_inr:.2f} / month")

    print("\n3. RATE LIMIT FEASIBILITY (PEAK TRAFFIC)")
    print(
        f"   • Estimated Peak RPM:   {peak_rpm:,.1f} / {provider_rpm_limit:,} RPM "
        f"({(peak_rpm/provider_rpm_limit)*100:.1f}%)"
    )
    print(
        f"   • Estimated Peak TPM:   {peak_tpm:,.0f} / {provider_tpm_limit:,} TPM "
        f"({(peak_tpm/provider_tpm_limit)*100:.1f}%)"
    )

    print("\n4. FINAL VERDICT")
    budget_status = (
        f"PASS (₹{budget_diff:,.2f} remaining | {budget_pct:.1f}% used)"
        if within_budget
        else f"FAIL (Over budget by ₹{abs(budget_diff):,.2f} | {budget_pct:.1f}% used)"
    )
    rpm_status = (
        "PASS"
        if within_rpm
        else f"FAIL (Exceeds limit by {math.ceil(peak_rpm - provider_rpm_limit):,} RPM)"
    )
    tpm_status = (
        "PASS"
        if within_tpm
        else f"FAIL (Exceeds limit by {math.ceil(peak_tpm - provider_tpm_limit):,} TPM)"
    )

    print(f"   • Budget Check (₹{monthly_budget_inr:,.2f}): {budget_status}")
    print(f"   • RPM Limit Check:             {rpm_status}")
    print(f"   • TPM Limit Check:             {tpm_status}")
    print("=" * 62)
```

```python
%%writefile budget_estimator.py
from dataclasses import dataclass
from typing import Dict


@dataclass
class BudgetReport:
    monthly_requests: float
    monthly_input_tokens: float
    monthly_output_tokens: float
    monthly_total_tokens: float
    total_cost_usd: float
    total_cost_inr: float
    peak_rpm: float
    peak_tpm: float
    within_budget: bool
    within_rpm: bool
    within_tpm: bool


def calculate_budget(
    daily_active_users: int,
    requests_per_user_per_day: float,
    avg_input_tokens: int,
    avg_output_tokens: int,
    in_price_usd_per_1m: float,
    out_price_usd_per_1m: float,
    usd_to_inr: float,
    provider_rpm_limit: int,
    provider_tpm_limit: int,
    monthly_budget_inr: float,
    days_per_month: int = 30,
    peak_traffic_multiplier: float = 3.0,
) -> BudgetReport:
    """Pure calculation function for LLM token usage, INR cost, and rate-limit checks."""
    daily_requests = daily_active_users * requests_per_user_per_day
    monthly_requests = daily_requests * days_per_month

    monthly_input_tokens = monthly_requests * avg_input_tokens
    monthly_output_tokens = monthly_requests * avg_output_tokens
    monthly_total_tokens = monthly_input_tokens + monthly_output_tokens

    input_cost_usd = (monthly_input_tokens / 1_000_000) * in_price_usd_per_1m
    output_cost_usd = (monthly_output_tokens / 1_000_000) * out_price_usd_per_1m
    total_cost_usd = input_cost_usd + output_cost_usd
    total_cost_inr = total_cost_usd * usd_to_inr

    minutes_per_day = 1440
    avg_rpm = daily_requests / minutes_per_day
    peak_rpm = avg_rpm * peak_traffic_multiplier

    tokens_per_req = avg_input_tokens + avg_output_tokens
    peak_tpm = peak_rpm * tokens_per_req

    return BudgetReport(
        monthly_requests=monthly_requests,
        monthly_input_tokens=monthly_input_tokens,
        monthly_output_tokens=monthly_output_tokens,
        monthly_total_tokens=monthly_total_tokens,
        total_cost_usd=total_cost_usd,
        total_cost_inr=total_cost_inr,
        peak_rpm=peak_rpm,
        peak_tpm=peak_tpm,
        within_budget=total_cost_inr <= monthly_budget_inr,
        within_rpm=peak_rpm <= provider_rpm_limit,
        within_tpm=peak_tpm <= provider_tpm_limit,
    )
```

```python
%%writefile test_budget.py
import unittest
from budget_estimator import calculate_budget


class TestLLMBudgetEstimator(unittest.TestCase):

    def setUp(self):
        """Standard baseline parameters for tests."""
        self.base_params = {
            "daily_active_users": 1000,
            "requests_per_user_per_day": 2,
            "avg_input_tokens": 1000,
            "avg_output_tokens": 500,
            "in_price_usd_per_1m": 1.00,
            "out_price_usd_per_1m": 4.00,
            "usd_to_inr": 90.0,
            "provider_rpm_limit": 100,
            "provider_tpm_limit": 200_000,
            "monthly_budget_inr": 20_000.0,
            "days_per_month": 30,
            "peak_traffic_multiplier": 3.0,
        }

    def test_token_and_inr_cost_calculation(self):
        """Verify exact token counts, USD cost, and INR conversion."""
        # 1000 users * 2 req * 30 days = 60,000 monthly requests
        # Input: 60,000 * 1,000 = 60M tokens -> 60 * $1.00 = $60.00
        # Output: 60,000 * 500 = 30M tokens -> 30 * $4.00 = $120.00
        # Total USD = $180.00 -> Total INR = 180 * 90.0 = ₹16,200.00
        report = calculate_budget(**self.base_params)

        self.assertEqual(report.monthly_requests, 60_000)
        self.assertEqual(report.monthly_input_tokens, 60_000_000)
        self.assertEqual(report.monthly_output_tokens, 30_000_000)
        self.assertEqual(report.monthly_total_tokens, 90_000_000)
        self.assertAlmostEqual(report.total_cost_usd, 180.00, places=2)
        self.assertAlmostEqual(report.total_cost_inr, 16_200.00, places=2)
        self.assertTrue(report.within_budget)

    def test_budget_exceeded(self):
        """Verify within_budget flips to False when INR cost exceeds budget."""
        params = {**self.base_params, "monthly_budget_inr": 15_000.0}
        report = calculate_budget(**params)

        self.assertAlmostEqual(report.total_cost_inr, 16_200.00, places=2)
        self.assertFalse(report.within_budget)

    def test_rate_limits_within_capacity(self):
        """Verify RPM and TPM pass when peak traffic fits provider limits."""
        # Daily requests = 2,000 -> Avg RPM = 2000 / 1440 = 1.3889
        # Peak RPM = 1.3889 * 3.0 = 4.1667 RPM (Limit: 100)
        # Peak TPM = 4.1667 * 1500 tokens = 6,250 TPM (Limit: 200,000)
        report = calculate_budget(**self.base_params)

        self.assertAlmostEqual(report.peak_rpm, 4.1667, places=3)
        self.assertAlmostEqual(report.peak_tpm, 6_250.0, places=1)
        self.assertTrue(report.within_rpm)
        self.assertTrue(report.within_tpm)

    def test_rpm_and_tpm_limits_exceeded(self):
        """Verify RPM and TPM fail when traffic spikes beyond provider limits."""
        params = {
            **self.base_params,
            "daily_active_users": 50_000,  # 100,000 daily req -> ~208.33 peak RPM
            "provider_rpm_limit": 200,
            "provider_tpm_limit": 250_000,  # 208.33 * 1500 = 312,500 peak TPM
        }
        report = calculate_budget(**params)

        self.assertFalse(report.within_rpm)
        self.assertFalse(report.within_tpm)

    def test_zero_traffic_edge_case(self):
        """Verify zero active users produces zero cost and passes all checks."""
        params = {**self.base_params, "daily_active_users": 0}
        report = calculate_budget(**params)

        self.assertEqual(report.monthly_requests, 0)
        self.assertEqual(report.total_cost_inr, 0.0)
        self.assertEqual(report.peak_rpm, 0.0)
        self.assertTrue(report.within_budget)
        self.assertTrue(report.within_rpm)
        self.assertTrue(report.within_tpm)


if __name__ == "__main__":
    unittest.main(argv=["first-arg-is-ignored"], verbosity=2)
```
